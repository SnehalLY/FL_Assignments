### **Implementation of Federated Learning using Federated Averaging on Linear Regression Model**




* **Introduction:** Federated Learning is a decentralized machine learning approach where multiple clients collaboratively train a model without sharing their local data. This ensures data privacy and reduces data transfer.


* **Methodology:** In this assignment, a federated learning system is simulated using multiple clients and a central server. Each client trains a local linear regression model using its own dataset. The server aggregates the model weights using the Federated Averaging algorithm.


* **Working:** Synthetic datasets are generated for each client
Each client performs local training using gradient descent
The server collects weights and computes their average
The updated global model is redistributed
This process is repeated for multiple rounds

* **Result:** The global model gradually converges towards the true relationship (y = 2x), demonstrating the effectiveness of federated learning.

* **Conclusion:** This assignment successfully demonstrates how federated learning enables collaborative model training without sharing raw data. It highlights the importance of distributed learning and privacy preservation in modern AI systems.

```python
import numpy as np

# --------------------------------------------------------
# CLIENT DEFINITION
# --------------------------------------------------------
class Client:
    def __init__(self, client_id, data, labels, lr=0.01):
        self.client_id = client_id
        self.data = data
        self.labels = labels
        self.lr = lr
        self.weights = np.random.randn(data.shape[1])  # simple linear model

    def train_local(self, epochs=5):
        """
        Local training using simple gradient descent for linear regression
        """
        for _ in range(epochs):
            predictions = self.data.dot(self.weights)
            errors = predictions - self.labels
            gradient = self.data.T.dot(errors) / len(self.data)
            self.weights -= self.lr * gradient

        print(f"Client {self.client_id} finished local training.")
        return self.weights


# --------------------------------------------------------
# SERVER DEFINITION
# --------------------------------------------------------
class Server:
    def __init__(self, num_clients):
        self.num_clients = num_clients
        self.global_weights = None

    def aggregate(self, client_weights):
        """
        Federated averaging of all client model parameters
        """
        self.global_weights = np.mean(client_weights, axis=0)
        print("\nServer aggregated weights using Federated Averaging...\n")
        return self.global_weights

    def send_weights(self):
        return self.global_weights


# --------------------------------------------------------
# SIMULATION SETUP
# --------------------------------------------------------
def generate_client_data(num_clients=3, samples_per_client=30):
    """
    Generate synthetic data for each client.
    y = 2x + noise
    """
    clients = []
    for i in range(num_clients):
        x = np.random.rand(samples_per_client, 1)
        y = 2 * x.squeeze() + np.random.randn(samples_per_client) * 0.1
        clients.append((x, y))
    return clients


def federated_learning_simulation(num_clients=3, rounds=5):
    # Create server
    server = Server(num_clients)

    # Generate client data
    client_data = generate_client_data(num_clients)

    # Initialize clients
    clients = [
        Client(i, data=x, labels=y)
        for i, (x, y) in enumerate(client_data)
    ]

    # GLOBAL MODEL INITIALIZATION
    server.global_weights = np.zeros(clients[0].weights.shape)

    for r in range(rounds):
        print(f"\n================ Federated Learning Round {r+1} ================\n")

        client_weights = []

        # SERVER sends global weights to all clients
        for client in clients:
            client.weights = server.send_weights().copy()

        # CLIENTS train on local data
        for client in clients:
            w = client.train_local()
            client_weights.append(w)

        # SERVER aggregates client updates
        global_w = server.aggregate(client_weights)

        print(f"Updated Global Weights: {global_w}\n")

    print("\nTraining completed!")
    print(f"Final Global Model Weights: {server.global_weights}")


if __name__ == "__main__":
    federated_learning_simulation()
```
